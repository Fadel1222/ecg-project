```python
import numpy as np
import matplotlib.pyplot as plt

fs = 1000
t = np.arange(0, 1, 1/fs)
x = np.sin(2*np.pi*5*t)

bits = 3
levels = 2**bits
xq = np.round((x + 1) / 2 * (levels - 1)) / (levels - 1) * 2 - 1

plt.plot(t, x, label='original')
plt.step(t, xq, label=f'{bits}-bit quantized', where='mid')
plt.legend()
plt.show()
```

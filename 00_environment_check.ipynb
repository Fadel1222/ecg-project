```python
import subprocess
from pathlib import Path

project_root = Path(r"D:\The Forge\ecg-project")
git_executable = r"C:\Program Files\Git\cmd\git.exe"
gitignore_file = project_root / ".gitignore"

# Read the original tracked version without changing Git history.
result = subprocess.run(
    [git_executable, "-C", str(project_root), "show", "HEAD:.gitignore"],
    capture_output=True,
    encoding="utf-8",
    check=True,
)

# Preserve our current rules as well.
current_rules = gitignore_file.read_text(encoding="utf-8")

combined_rules = (
    result.stdout.rstrip()
    + "\n\n# ECG project additions\n"
    + current_rules
)

gitignore_file.write_text(combined_rules, encoding="utf-8")

print("Original and ECG-specific ignore rules combined.")

# Show the remaining changes with explicit UTF-8 decoding.
review = subprocess.run(
    [git_executable, "-C", str(project_root), "diff", "--", ".gitignore"],
    capture_output=True,
    encoding="utf-8",
    check=True,
)

print(review.stdout)
```

```python
import subprocess

git_executable = r"C:\Program Files\Git\cmd\git.exe"
project_path = r"D:\The Forge\ecg-project"

setup_files = [
    ".gitignore",
    "README.md",
    "environment.yml",
    "docs/project_specification.md",
    "00_environment_check.ipynb",
]

subprocess.run(
    [git_executable, "-C", project_path, "add", "--", *setup_files],
    capture_output=True,
    encoding="utf-8",
    check=True,
)

for arguments in [
    ["diff", "--cached", "--stat"],
    ["status", "--short"],
]:
    result = subprocess.run(
        [git_executable, "-C", project_path, *arguments],
        capture_output=True,
        encoding="utf-8",
        check=True,
    )

    print(f"\n--- git {' '.join(arguments)} ---")
    print(result.stdout)
```
